# 📑 Annual Data Merging & Data exploration (2021 - 2024)

Our first goal is to consolidate the four core files (`caract`, `lieux`, `vehicules`, `usagers`) into a single, comprehensive dataset for each individual year. 

### Merging Logic:
1. Connect **Vehicles** and **Users** using the combined key: `Num_Acc` + `num_veh`.
2. Append **Places** using the primary key: `Num_Acc`.
3. Append **Characteristics**. Account for the **2022 anomaly**, where `Num_Acc` was labeled as `Accident_Id` in the characteristics file.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

```python
df_users=pd.read_csv("../../data_acc/raw/usagers-2021.csv", sep=";")
df_places=pd.read_csv("../../data_acc/raw/lieux-2021.csv", sep=";", header=0, low_memory=False)
df_caract=pd.read_csv("../../data_acc/raw/carcteristiques-2021.csv", sep = ";", encoding='utf-8')
df_veh=pd.read_csv("../../data_acc/raw/vehicules-2021.csv", sep=";")
```

```python
# 1. First join: Users and Vehicles 
df_merged = pd.merge(df_users, df_veh, on=["Num_Acc", "num_veh"], how="inner")

# 2. Second join: Add road infrastructure data (Places)
df_merged = pd.merge(df_merged, df_places, on="Num_Acc", how="left")

# 3. Third join: Add environmental details (Characteristics)
df_merged_21 = pd.merge(df_merged, df_caract, on="Num_Acc", how="left")

print("📊 Merging successful!")
print(f"Consolidated 2021 Dataset Shape: {df_merged_21.shape}")
```

```python
#building df_merged_2022

df_users=pd.read_csv("../../data_acc/raw/usagers-2022.csv", sep=";")
df_places=pd.read_csv("../../data_acc/raw/lieux-2022.csv", sep=";", header=0, low_memory=False)
df_caract=pd.read_csv("../../data_acc/raw/carcteristiques-2022.csv", sep = ";", encoding='utf-8')
df_veh=pd.read_csv("../../data_acc/raw/vehicules-2022.csv", sep=";")

if 'Accident_Id' in df_caract.columns:
    df_caract.rename(columns={'Accident_Id': 'Num_Acc'}, inplace=True)
    print("-> Anomaly detected & fixed: Renamed 'Accident_Id' to 'Num_Acc'")

# 1. First join: Users and Vehicles 
df_merged = pd.merge(df_users, df_veh, on=["Num_Acc", "num_veh"], how="inner")

# 2. Second join: Add road infrastructure data (Places)
df_merged = pd.merge(df_merged, df_places, on="Num_Acc", how="left")

# 3. Third join: Add environmental details (Characteristics)
df_merged_22 = pd.merge(df_merged, df_caract, on="Num_Acc", how="left")

print("📊 Merging successful!")
print(f"Consolidated 2022 Dataset Shape: {df_merged_22.shape}")
```

## 📊 Target Distribution & Missing Data Consistency Analysis

After processing and merging the datasets for 2021, 2022, 2023, and 2024, we conducted a cross-year data quality check. 

### Key Findings:
1. **Target Stability**: The distribution of our target variable `grav` remains highly consistent across all years.
2. **Missing Data Patterns**: The exact same columns suffer from missing values (`NaN`) across the entire 2021-2024 period. 

During the exploration of the target variable `grav` (injury severity), a significant shift in data logging was discovered between the historical data and the latest 2024 dataset.

#### Key Observation:
* **Years 2021, 2022, 2023**: The dataset contains a value of `-1`. In the French BAAC system, `-1` stands for *"Non renseigné"* (Not specified / Unknown).
* **Year 2024**: The value `-1` completely disappears from the dataset.
